```python
import numpy as np
import pandas as pd
```

```python
netflix_watch_time = np.random.randint(10,601,1000)
netflix_watch_time
```

```python
watch_time = pd.Series(netflix_watch_time, name="watch_time_per_day")
watch_time
```

```python
#lowest
watch_time.min()
```

```python
#top 10 usage time
watch_time.nlargest(10)
```

```python
#user use above average
average = watch_time.mean()
average
```

```python
watch_time[watch_time >average]
```

```python
#median
watch_time.median()
```

```python
#convert minutes into hours
watch_time_hours = watch_time / 60
watch_time_hours
```

```python
#total usage time
watch_time_hours.sum()
```

```python
#more than 5 hours users
watch_time_hours[watch_time_hours > 5]
```

```python
#difference between highest and lowest usage
watch_time.max() - watch_time.min()
```

```python
# reduce users usage time by 10%
watch_time_reduction = watch_time * 0.9
w = watch_time_reduction.sum()
w
```

```python
# new average after 10% reduction
new_average = watch_time_reduction.mean()
new_average
```
