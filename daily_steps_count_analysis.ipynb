```python
import numpy as np
import pandas as pd
```

```python
daily_steps  = np.random.randint(1000,30000,(1500,7))
daily_steps
```

```python
daily_steps_count = pd.DataFrame(daily_steps)
name = ["sunday", "monday", "tuesday", "wednesday", "thursday", "friday", "saturday"]
daily_steps_count.columns = name
daily_steps_count
```

```python
daily_steps_count.size
```

```python
daily_steps_count.shape
```

```python
daily_steps_count.ndim
```

```python
daily_steps_count.index
```

```python
daily_steps_count.isnull().sum()
```

```python
#total steps
daily_steps_count.sum()
```

```python
#average
daily_steps_count.mean()
```

```python
#median
daily_steps_count.median()
```

```python
#highest steps count
daily_steps_count.max()
```

```python
#lowest steps count
daily_steps_count.min()
```

```python
#top 10 highest count
daily_steps_count.nlargest(10, columns=['sunday', 'monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday'])
```

```python
#bottom 10 highest count
daily_steps_count.nsmallest(10, columns=['sunday', 'monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday'])
```

```python
#count days more than 10000 steps
daily_steps_count[daily_steps_count > 10000]
```

```python
daily_steps_count[daily_steps_count > 10000].count()
```

```python
#steps less than 5000
daily_steps_count[daily_steps_count < 5000]
```

```python
daily_steps_count[daily_steps_count < 5000].count()
```

```python
#count days between 7000 to 12000
daily_steps_count[(daily_steps_count > 7000) & (daily_steps_count < 12000)]
```

```python
daily_steps_count[(daily_steps_count > 7000) & (daily_steps_count < 12000)].count()
```

```python
#percentage of days above 10000 steps
daily_steps_count[daily_steps_count > 10000].count()/daily_steps_count.count()*100
```

```python
#steps greater than average
daily_steps_count.mean()
```

```python
daily_steps_count[daily_steps_count > daily_steps_count.mean()]
```

```python
#steps less than median
daily_steps_count.median()
```

```python
daily_steps_count[daily_steps_count > daily_steps_count.median()]
```

```python
# index of the most active day
daily_steps_count.nlargest(1, columns=['sunday','monday','tuesday','wednesday','thursday','friday','saturday']).index
```

```python
# index of the most active day
daily_steps_count.nsmallest(1, columns=['sunday','monday','tuesday','wednesday','thursday','friday','saturday']).index
```

```python
#sort in ascending order
daily_steps_count.sort_values(by=['sunday','monday','tuesday','wednesday','thursday','friday','saturday'])
```

```python
#sort in descending order
daily_steps_count.sort_values(by=['sunday','monday','tuesday','wednesday','thursday','friday','saturday'], ascending=False)
```

```python
# difference between the most and least active days
daily_steps_count.max() - daily_steps_count.min()
```

```python
#increase steps by 15%
increase= daily_steps_count * 1.15
increase
```

```python
#new average
increase.mean()
```

```python
#cross 10000 steps after increase by 15%
increase[increase > 10000]
```
